```python
#Here i laid down the TPM of a single beethoven song
from collections import defaultdict

# Step 2: Create the transition probability matrix (TPM) for order 1 Markov process
def create_transition_matrix_order_1(notes):
    transition_matrix = defaultdict(lambda: defaultdict(int))
    for i in range(len(notes) - 1):
        current_note = notes[i]
        next_note = notes[i + 1]
        transition_matrix[current_note][next_note] += 1
    
    # Convert counts to probabilities
    for current_note in transition_matrix:
        total_transitions = sum(transition_matrix[current_note].values())
        for next_note in transition_matrix[current_note]:
            transition_matrix[current_note][next_note] /= total_transitions
    
    return transition_matrix

# Step 3: Create the transition probability matrix (TPM) for order 2 Markov process
def create_transition_matrix_order_2(notes):
    transition_matrix = defaultdict(lambda: defaultdict(int))
    for i in range(len(notes) - 2):
        current_notes = (notes[i], notes[i + 1])
        next_note = notes[i + 2]
        transition_matrix[current_notes][next_note] += 1
    
    # Convert counts to probabilities
    for current_notes in transition_matrix:
        total_transitions = sum(transition_matrix[current_notes].values())
        for next_note in transition_matrix[current_notes]:
            transition_matrix[current_notes][next_note] /= total_transitions
    
    return transition_matrix

# Create TPM for order 1 Markov process
tpm_order_1 = create_transition_matrix_order_1(notes)
print("\nTransition Probability Matrix (Order 1):")
for current_note, next_notes in tpm_order_1.items():
    print(current_note, next_notes)

# Create TPM for order 2 Markov process
tpm_order_2 = create_transition_matrix_order_2(notes)
print("\nTransition Probability Matrix (Order 2):")
for current_notes, next_notes in tpm_order_2.items():
    print(current_notes, next_notes)
```
